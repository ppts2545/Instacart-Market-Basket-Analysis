💡 คำถามสำคัญที่ Poom ต้องตอบเพื่อปิดโปรเจกต์:

    Customer Quality: ลูกค้าใหม่ที่เข้ามา (New Comers) มี avg_item_weight และ basket_size สูงกว่าคนเก่าที่หายไปไหม?

        ถ้าคนใหม่ซื้อน้อยกว่า แต่คนเก่าที่หายไปเคยซื้อเยอะ = สัญญาณอันตราย (คุณภาพลูกค้าลดลง)

    The Churn Impact: มูลค่าความเสียหายจากกลุ่ม Churned ที่เราวิเคราะห์ไป (7.7 ล้าน) เมื่อหารเฉลี่ยต่อเดือนแล้ว มันสูงกว่ายอดขายที่ได้จากลูกค้าใหม่ในเดือนล่าสุดหรือไม่?

    Growth Velocity: อัตราการเพิ่มของ total_value มันเร็วกว่าอัตราการเพิ่มของ total_users หรือไม่? (ถ้า User เพิ่มเร็วกว่า Revenue แปลว่าเราได้แต่ "ปริมาณ" แต่ไม่ได้ "คุณภาพ")
